```python
#Creating a String
first_name = "Rahul"
last_name = "Patel"
# Concatenation (Joining Strings)
full_name= first_name +" "+ last_name 
print("Full Name:", full_name)
```

```python
#Accessing Characters
print("First Character:", full_name[0]) 
print("Last Character:", full_name[-1])
```

```python
# Slicing (Extracting Part of String)
print("First Name:", full_name[0:5])

print("Last Name:", full_name[6:])
```

```python
# Searching Text
print("Contains 'Patel':", "Patel" in full_name)
print("Position of 'Patel':", full_name.find("Patel"))
```

```python
# Replacing Text
new_name=full_name.replace("Rahul", "Rohan")
print("After Replace:", new_name)
```

```python
# Changing Case
print("Uppercase:", full_name.upper())
print("Lowercase:",full_name.lower())
print("Title Case:", full_name.title())
```

```python
# Removing Extra Spaces
text =" Welcome to Python Programming"
print("Original:", text)
print("After Removing Spaces:", text.strip())
```
