```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#monthly sales data
month=["Jan","Feb","Mar","Apr","May","Jun"]
sales=[120000,150000,135000,180000,210000,195000]

#product-wise revenue
products=["Laptop","Mobile","Tablet","Headphones"]
revenue=[450000,350000,180000,120000]

#customer data
customer=[120,150,135,180,210,195]
```

```python
#1.Line chart
plt.plot(month,sales,marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales performance")
```

```python
#2.Bar chart
plt.bar(products,revenue)
plt.xlabel("products")
plt.ylabel("revenue")
plt.title("Products Revenue performance")
```

```python
#2.Barplot chart
sns.barplot(x=products,y=revenue)
plt.xlabel("products")
plt.ylabel("revenue")
plt.title("Product-Wise Revenue")
```

```python
#3.Scatter chart
plt.scatter(customer,sales)
plt.xlabel("customer")
plt.ylabel("sales")
plt.title("Customer Count vs sales")
```

```python
#4.Histogram
plt.hist(sales,bins=5)
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.title("sales Distribution")
```

```python
#5.Pie chart
plt.pie(revenue,labels=products,autopct="%1.1f%%")
plt.title("Product-Wise Revenue Contribution")
```

```python
#Line chart
sns.lineplot(x=month,y=sales,marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales performance")
```

```python
#Scatter chart
sns.scatterplot(x=customer,y=sales)
plt.xlabel("customer")
plt.ylabel("sales")
plt.title("Customer Count vs sales")
```
