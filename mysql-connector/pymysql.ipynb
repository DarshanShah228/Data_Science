```python
import pymysql

# Connect to MySQL Server
conn = pymysql.connect(
    host="localhost",
    user="root",
    password=""
)

print("MySQL Server Connected Successfully!")

cursor = conn.cursor()

# Create Database
cursor.execute("CREATE DATABASE IF NOT EXISTS university")
print("Database 'university' created successfully!")

# Select Database
cursor.execute("USE university")
print("University database selected!")

# Create Table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS student_details (
        id INT AUTO_INCREMENT PRIMARY KEY,
        name VARCHAR(50),
        course VARCHAR(50),
        marks INT
    )
""")

print("Student Details Table Created Successfully!")

# Insert Student Records
sql = """
    INSERT INTO student_details (name, course, marks)
    VALUES (%s, %s, %s)
"""

students = [
    ("Darshan", "MCA", 85),
    ("Meet", "MCA", 72),
    ("Abhay", "MCA", 99),
    ("Nikhil", "MCA", 55)
]

cursor.executemany(sql, students)

# Save Changes
conn.commit()

print("Student records inserted successfully!")

# Display Student Records
cursor.execute("SELECT * FROM student_details")

records = cursor.fetchall()

print("\nStudent Records:")
print("------------------------")

for row in records:
    print(row)

# Close Connection
cursor.close()
conn.close()

print("\nMySQL connection closed.")
```
