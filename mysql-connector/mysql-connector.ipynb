```python
#create database
import mysql

conn = mysql.connector.connect(
host="localhost",
user="root",
password=""
)
print("MySQL Server Connected Successfully!")
cursor = conn.cursor()
cursor.execute("CREATE DATABASE IF NOT EXISTS college")
print("Database 'college' created successfully!")
cursor.execute("USE college")
print("College database selected!")

cursor.execute("""
CREATE TABLE IF NOT EXISTS student (
id INT AUTO_INCREMENT PRIMARY KEY,
name VARCHAR(50),
course VARCHAR(50),
marks INT
)
""")
print("Student Table Created Successfully!")
```

```python
sql="""
INSERT INTO student(name,course,marks)
VALUES(%s,%s,%s)
"""
student=[
    ("Darshan","MCA",85),
    ("Meet","MCA",72),
    ("Abhay","MCA",99),
    ("Nikhil","MCA",55),
]

cursor.executemany(sql,student)

#save
conn.commit()
print("student record inserted successfully!!")
```

```python
cursor.execute("SELECT * FROM student")
records=cursor.fetchall()

print("\nStudent Records:")
print("------------------------")

for row in records:
    print(row)

#close 
cursor.close()
```
