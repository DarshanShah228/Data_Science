```python
import sqlite3
import pandas as pd

# Connect to a temporary, in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create table
cursor.execute('''
CREATE TABLE Sales (
    SaleID INTEGER PRIMARY KEY AUTOINCREMENT,
    CustomerName TEXT NOT NULL,
    Product TEXT NOT NULL,
    Category TEXT NOT NULL,
    City TEXT NOT NULL,
    Quantity INTEGER NOT NULL,
    Price REAL NOT NULL
);
''')

# Insert sample dataset
mock_data = [
    ('Aarav Mehta', 'Laptop', 'Electronics', 'Ahmedabad', 1, 55000.00),
    ('Neha Sharma', 'Smartphone', 'Electronics', 'Mumbai', 3, 15000.00),
    ('Rahul Patel', 'Desk Chair', 'Furniture', 'Ahmedabad', 4, 4500.00),
    ('Priya Nair', 'Headphones', 'Electronics', 'Bangalore', 2, 2500.00),
    ('Amit Verma', 'Coffee Maker', 'Appliances', 'Ahmedabad', 1, 8000.00),
    ('Siddharth Shah', 'Smart Watch', 'Electronics', 'Ahmedabad', 5, 12000.00),
    ('Ananya Rao', 'Dining Table', 'Furniture', 'Delhi', 1, 22000.00),
    ('Vikram Singh', 'Microwave', 'Appliances', 'Mumbai', 2, 9500.00)
]

cursor.executemany('''
INSERT INTO Sales (CustomerName, Product, Category, City, Quantity, Price)
VALUES (?, ?, ?, ?, ?, ?)
''', mock_data)

conn.commit()
print("Database created successfully!")
```

```python
#22. Display each Category, the number of sales, total quantity, minimum price, and maximum price, grouped by Category and sorted by total quantity in descending order.
cursor.execute("""
    SELECT Category, COUNT(*), SUM(Quantity), MIN(Price), MAX(Price) 
    FROM Sales 
    GROUP BY Category 
    ORDER BY SUM(Quantity) DESC;
""")
for row in cursor.fetchall():
    print(row)
```
