```python

import mysql.connector as conn
```

```python
conn=conn.connect(
    host="localhost",
    username="root",
    password="",
    port=3309,
    database="Students"
)
print("Connect")
```

```python
cur = conn.cursor()
# Execute the SQL statement
cur.execute(
    """
    CREATE TABLE IF NOT EXISTS students (
        id INT PRIMARY KEY,
        name VARCHAR(50),
        age INT,
        course VARCHAR(50)
    );
"""
)
print("Table reated successfully!")
```

```python
sql = "INSERT INTO students (id, name, age, course) VALUES (%s, %s, %s, %s)"

values = [
    (1, "Rahul", 22, "MCA"), 
    (2, "Priya", 23, "MCA"), 
    (4, "Amit", 21, "MCA")
]

cur.executemany(sql, values)
conn.commit()

print("Records inserted successfully!")
```

```python
sql = "UPDATE students SET age=%s WHERE id=%s" 
value = (21, 2) 
cur.execute(sql, value) 
conn.commit() 
print("Record updated successfully!") 

sql = "DELETE FROM students WHERE id = %s" 
value = (3,)
cur.execute(sql, value)
conn.commit()
print("Record delete successfully!")
```

```python
cur.execute("""
    ALTER TABLE students
    ADD email VARCHAR(100)
""")
print("Column added successfully!")
```

```python
cur.execute("TRUNCATE TABLE students")
print("All records deleted using TRUNCATE!")

cur.execute("DROP TABLE students")
print("Table dropped successfully!")

cur.close()
conn.close()
print("MySQL connection closed!")
```
