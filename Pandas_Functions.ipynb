```python
import pandas as pd
import numpy as np
```

```python
#Create DataFrame
df=pd.DataFrame({
"EmpID": [101, 102, 103, 104, 105, 105, 106],
"Name": ["Amit", "Riya", "Neha", "Raj", "Karan", "Karan", "Priya"], 
"Department": ["IT", "HR", "IT", "Sales", "HR", "HR", "IT"], 
"City": ["Ahmedabad", "Surat", "Rajkot", "Ahmedabad", np.nan, "Surat", "Vadodara"], 
"Salary": [50000, 45000, 70000, 35000, np.nan, 52000, 65000],
"Grade": ["C", "B", "A", "С", "В", "B", "A"]
})

print("Original DataFrame")
print(df)
```

```python
print("\nNumber of Employees in Each Department:\n")
print(df["Department"].value_counts())
```

```python
print("\nUnique Cities:\n")
print(df["City"].unique())
```

```python
print("\nTotal Unique Department:")
print(df["Department"].nunique())
```

```python
print("\nDuplicate Record:")
print(df.duplicated())
```

```python
df=df.drop_duplicates()
```

```python
print("\n After Removing Duplicate:\n")
print(df)
```

```python
print("\nMissing Values:")
print(df.isnull())
```

```python
print("\nTotal Missing Values")
print(df.isnull().sum())
```

```python
df["City"] = df["City"].fillna ("Unknown")
df ["Salary"] = df["Salary"].fillna (df["Salary"].mean())
```

```python
print("\nAfter Filling Missing Values")
print(df)
```

```python
df["Salary"] = df ["Salary"].astype(int)
print("\nAfter Changing Data Type")
print(df.dtypes)
```

```python
print("\nRandom 3 Employees")
print(df.sample(5))
```

```python
print("\nTop 3 Highest Salaries")
print(df.nlargest (3, "Salary"))
```

```python
print("\nBottom 2 Lowest Salaries")
print(df.nsmallest(2, "Salary"))
```

```python
grade={
    'A':"Excellent",
    'B':"Good",
    'C':"Average",
    'D':"Bad"
}
print(df)
df["Performance"]=df["Grade"].map(grade)
print("Final Dataframe:\n")
print(df)
```
