```python
import pandas as pd
```

```python
# Create ShopKart customer dataset
data = {
    "Customer": ["Amit", "Priya", "Rahul", "Neha", "Karan",
                 "Pooja", "Ravi", "Meera", "Jay", "Riya"],
                 
    "Age":[22,35,28,42,19,31,50,26,38,45],
    
    "Spending":[15000,45000,28000,75000,12000,
               52000,90000,22000,65000,38000],
                 
    "Discount": [5, 15, 10, 25, 5, 20, 30, 10, 25, 15]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
# Define the custom function to categorize ages
def age_group(age):
    if age < 25:
        return "Young"
    elif age < 40:
        return "Adult"
    else:
        return "Senior"

# Apply the function to create a new column
df["Age_Group"] = df["Age"].apply(age_group)

# Print the result
print("\nAfter apply() - Age Group:")
print(df)
```

```python
discount_category = {
    5: "Low",
    10: "Low",
    15: "Medium",
    20: "Medium",
    25: "High",
    30: "High"
}

df["Discount_Category"]=df ["Discount"].map(discount_category)
print("\nAfter map() Discount Category:")
print(df)
```

```python
df["Final_Spending"] = df.apply(
lambda row: row ["Spending"]-(row ["Spending"] * row["Discount"] / 100), 
    axis=1)

print("\nAfter lambda Final Spending:")
print(df)

df["Spending_Level"] = pd.cut(
    df ["Spending"],
    bins=[0, 20000, 50000, 100000], 
    labels=["Low", "Medium", "High"]
)
```
