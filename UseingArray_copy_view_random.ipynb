```python
import numpy as np
```

```python
#1. Create an array
arr=np.array ([10, 20, 30, 40, 50])
print("Original Array:")
print(arr)
```

```python
#2. Copy
copy_arr = arr.copy()
copy_arr[0] = 100
print("\nAfter modifying Copy:")
print("Original Array:", arr)
print("Copy Array:", copy_arr)
```

```python
#3. View
view_arr=arr.view()
view_arr[1] = 200
print("\nAfter modifying View:")
print("Original Array:", arr)
print("View Array:", view_arr)
```

```python
#4. Random Numbers
random_arr = np.random.randint(1, 100, 5)
print("\nRandom Array:")
print(random_arr)
```

```python
#5. Searching
position= np.where (random_arr > 50)
print("\nPositions of numbers greater than 50:")
print(position)
```
