```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Create ShopKart customer dataset
data = {
    "Customer": ["Amit", "Priya", "Rahul", "Neha", "Karan",
                 "Pooja", "Ravi", "Meera", "Jay", "Riya"],
                 
    "Product": ["Laptop", "Mobile", "Headphones", "Laptop", "Mobile", 
                "Tablet", "Headphones", "Mobile", "Laptop", "Tablet"],

     "Category": ["Electronics", "Electronics", "Accessories", "Electronics", "Electronics", 
                 "Electronics", "Accessories", "Electronics", "Electronics", "Electronics"],
    
    "Month": ["Jan", "Jan", "Feb", "Feb", "Mar", 
              "Mar", "Apr", "Apr", "May", "May"],
    
    "Sales":[15000,45000,28000,75000,12000,
               52000,90000,22000,65000,38000],
                 
    "Quantity": [1, 2, 3, 1, 2, 
                 1, 4, 2, 1, 1]
}

df = pd.DataFrame(data)


print("Shopkart Sales Dataset:")
print(df)
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(
    df[["Sales","Quantity"]],
    annot=True,
    cmap="Blues",
    linewidths=0.5
)
plt.title("Monthly Sales Breakdown by Product")
plt.show()
```
