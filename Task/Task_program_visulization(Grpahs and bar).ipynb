```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Month': [
        'January', 'February', 'March', 'April',
        'May', 'June', 'July', 'August',
        'September', 'October', 'November', 'December'
    ],

    'Product': [
        'Laptop', 'Mobile', 'Headphones', 'Laptop',
        'Mobile', 'Headphones', 'Laptop', 'Mobile',
        'Headphones', 'Laptop', 'Mobile', 'Headphones'
    ],

    'Sales': [
        50000, 35000, 15000, 55000,
        40000, 18000, 60000, 45000,
        20000, 65000, 50000, 22000
    ],

    'Quantity': [
        10, 25, 30, 11,
        28, 35, 12, 30,
        40, 13, 32, 42
    ],

    'Customers': [
        8, 20, 25, 9,
        23, 30, 10, 25,
        35, 11, 27, 37
    ]
}

df = pd.DataFrame(data)

print("ShopKart Sales Dataset:")
print(df)
```

```python
# 1. Line chart - Monthly Sales
plt.figure(figsize=(10, 5))
plt.plot(df['Month'], df['Sales'], marker='o', color='blue')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 2. Bar chart - Product-wise Sales
product_sales = df.groupby('Product')['Sales'].sum()

plt.figure(figsize=(7, 5))
plt.bar(product_sales.index, product_sales.values,
        color=['orange', 'green', 'purple'])
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Product-wise Sales')
plt.show()
```

```python
# 3. Scatter plot - Customers vs Sales
plt.figure(figsize=(7, 5))
plt.scatter(df['Customers'], df['Sales'],
            color='red', s=80)
plt.xlabel('Number of Customers')
plt.ylabel('Sales')
plt.title('Customers vs Sales')
plt.grid(True)
plt.show()
```

```python
# 4. Histogram - Sales Distribution
plt.figure(figsize=(7, 5))
plt.hist(df['Sales'], bins=6, color='skyblue',
         edgecolor='black')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Sales Distribution')
plt.show()
```

```python
# 5. Pie chart - Product-wise Sales Contribution
plt.figure(figsize=(7, 7))
plt.pie(
    product_sales.values,
    labels=product_sales.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Product-wise Sales Contribution')
plt.show()
```

```python
sns.set_theme(style="whitegrid")

# 6. Seaborn Line Plot - Sales Trend
plt.figure(figsize=(10, 5))
sns.lineplot(
    data=df,
    x='Month',
    y='Sales',
    marker='o',
    color='blue'
)
plt.title('Sales Trend')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 7. Seaborn Bar Plot - Average Sales by Product
plt.figure(figsize=(7, 5))

ax = sns.barplot(
    data=df,
    x='Product',
    y='Sales',
    hue='Product',
    palette='viridis'
)

# Hide legend manually
if ax.get_legend() is not None:
    ax.get_legend().remove()

plt.xlabel('Product')
plt.ylabel('Average Sales')
plt.title('Average Sales by Product')
plt.show()
```

```python
# 8. Seaborn Count Plot - Number of Purchases by Product
plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    x='Product',
    hue='Product',
    palette='Set2',
)
plt.xlabel('Product')
plt.ylabel('Number of Purchases')
plt.title('Number of Purchases for Each Product')
plt.show()
```

```python
# 9. Seaborn Box Plot - Distribution of Sales by Product
plt.figure(figsize=(7, 5))
sns.boxplot(
    data=df,
    x='Product',
    y='Sales',
    hue='Product',
    palette='Set3'
)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Distribution of Sales by Product')
plt.show()
```
