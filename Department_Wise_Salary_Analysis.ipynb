```python
import pandas as pd

df=pd.DataFrame({
    "Product_ID":[1,2,3,4,5,6],
    "ProductName":["A","B","C","D","E","F"],
    "Catagory":["Electronics","Clothing","Clothing","Electronics","Oil","Clothing"],
    "Branch":["Ahmedabad","Bhavnagar","Bhavnagar","Bhavnagar","Ahmedabad","Surat"],
    "City":["Ahmd","Ahmd","Palitana","Palitana","Palitana","Palitana"],
    "Sales":[12500,33500,56400,32300,55000,70000],
    "Profit":[1250,1700,5000,7500,10000,200]
})

df
```
