```python
from datetime import datetime , timedelta

# Get current date and time
now = datetime.now()

# Format as DD-Mon-YYYY HH:MM AM/PM
formatted_now = now.strftime("%d-%b-%Y %I:%M %p")
print(formatted_now)
```

```python
dob = input("Enter Date of Birth (DD/MM/YYYY): ")

birth_date = datetime.strptime(dob, "%d/%m/%Y")
today = datetime.today()

years = today.year - birth_date.year
months = today.month - birth_date.month
days = today.day - birth_date.day

if days < 0:
    months -= 1
    days += 30

if months < 0:
    years -= 1
    months += 12

print("Age:", years, "Years", months, "Months", days, "Days")
```

```python
date1 = input("Enter first date (DD/MM/YYYY): ")
date2 = input("Enter second date (DD/MM/YYYY): ")

d1 = datetime.strptime(date1, "%d/%m/%Y")
d2 = datetime.strptime(date2, "%d/%m/%Y")

difference = abs(d2 - d1)

days = difference.days
weeks = days // 7
months = days // 30

print("Days:", days)
print("Weeks:", weeks)
print("Full Months:", months)
```

```python
start = input("Enter start date (DD/MM/YYYY): ")
end = input("Enter end date (DD/MM/YYYY): ")

start_date = datetime.strptime(start, "%d/%m/%Y")
end_date = datetime.strptime(end, "%d/%m/%Y")

current = start_date

print("Sundays:")

while current <= end_date:
    if current.weekday() == 6:
        print(current.strftime("%d-%m-%Y"))

    current += timedelta(days=1)
```

```python
from datetime import datetime

event = input("Enter event date and time (DD/MM/YYYY HH:MM): ")

event_time = datetime.strptime(event, "%d/%m/%Y %H:%M")

now = datetime.now()

remaining = event_time - now

days = remaining.days
hours = remaining.seconds // 3600
minutes = (remaining.seconds % 3600) // 60

print("Remaining Time:")
print(days, "Days")
print(hours, "Hours")
print(minutes, "Minutes")
```

```python
def is_leap_year(year):
    try:
        datetime(year, 2, 29)
        return True
    except ValueError:
        return False


year = int(input("Enter year: "))

print("Leap Year:", is_leap_year(year))
```

```python
joining = input("Enter joining date (DD/MM/YYYY): ")

join_date = datetime.strptime(joining, "%d/%m/%Y")

today = datetime.today()

anniversary = datetime(today.year, join_date.month, join_date.day)

if anniversary < today:
    anniversary = datetime(today.year + 1, join_date.month, join_date.day)

remaining = anniversary - today

print("Next Anniversary:", anniversary.strftime("%d-%m-%Y"))
print("Days Remaining:", remaining.days)
```

```python
from datetime import datetime

logs = [
    "2026-08-05 10:30:00",
    "2026-08-05 09:15:00",
    "2026-08-05 12:45:00"
]

dates = []

for log in logs:
    dates.append(datetime.strptime(log, "%Y-%m-%d %H:%M:%S"))

dates.sort()

print("Sorted Logs:")

for i in range(len(dates)):
    print(dates[i])

    if i > 0:
        gap = dates[i] - dates[i-1]
        print("Gap:", gap)
```

```python
from datetime import datetime, timedelta


def working_days(start, end, holidays):

    count = 0

    current = start

    while current <= end:

        if current.weekday() < 5 and current not in holidays:
            count += 1

        current += timedelta(days=1)

    return count


start = datetime.strptime("01/08/2026", "%d/%m/%Y")
end = datetime.strptime("10/08/2026", "%d/%m/%Y")


holidays = [
    datetime.strptime("05/08/2026", "%d/%m/%Y")
]


print("Working Days:",
      working_days(start, end, holidays))
```

```python
from datetime import datetime
from zoneinfo import ZoneInfo


meeting = input("Enter meeting time (YYYY-MM-DD HH:MM): ")


india_time = datetime.strptime(
    meeting,
    "%Y-%m-%d %H:%M"
)


india_time = india_time.replace(
    tzinfo=ZoneInfo("Asia/Kolkata")
)


new_york = india_time.astimezone(
    ZoneInfo("America/New_York")
)


london = india_time.astimezone(
    ZoneInfo("Europe/London")
)


print("India Time:",
      india_time.strftime("%d-%m-%Y %I:%M %p"))

print("New York:",
      new_york.strftime("%d-%m-%Y %I:%M %p"))

print("London:",
      london.strftime("%d-%m-%Y %I:%M %p"))
```
