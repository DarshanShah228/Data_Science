```python
import pandas as pd
import numpy as np
```

```python
# Create a mock employee dataset for demonstration
data = {
    'Emp_ID':[101,102,103,104,105,106,107,108,109],
    'Emp_Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Alice', 'Frank', 'Grace', 'Henry'],
    'Dept': ['HR', 'IT', 'Finance', 'IT', 'HR', 'HR', 'Finance', np.nan, 'Marketing'],
    'Salary_USD': ['50000', '75000', '60000', '80000', '52000', '50000', '95000', '45000', '70000'],
    'Grade': ['A', 'B', 'C', 'A', 'B', 'A', 'A', 'C', 'D']
}

df = pd.DataFrame(data)
print("--- Original Dataset ---")
print(df, "\n")
```

```python
# 1. Count employees department-wise
print("--- 1. Department-wise Employee Count ---")
dept_counts = df['Dept'].value_counts(dropna=False)
print(dept_counts, "\n")
```

```python
# 2. Find unique departments and count the number of unique departments
print("--- 2. Unique Departments ---")
unique_depts = df['Dept'].dropna().unique()
num_unique_depts = df['Dept'].nunique()
print(f"Unique Departments: {unique_depts}")
print(f"Number of Unique Departments: {num_unique_depts}\n")
```

```python
# 3. Identify and remove duplicate employee records
print("--- 3. Removing Duplicates ---")
duplicates = df[df.duplicated()]
print(f"Duplicate records found:\n{duplicates}")
df = df.drop_duplicates().reset_index(drop=True)
print("Dataset after removing duplicates:")
print(df, "\n")
```

```python
# 4. Identify and handle missing values
print("--- 4. Handling Missing Values ---")
print("Missing values per column:")
print(df.isnull().sum())
# Filling missing department values with 'Unknown'
df['Dept'] = df['Dept'].fillna('Unknown')
print("Dataset after filling missing values:")
print(df, "\n")
```

```python
# 5. Rename a column and change its data type
print("--- 5. Rename Column and Change Data Type ---")
# Rename 'Salary_USD' to 'Salary'
df = df.rename(columns={'Salary_USD': 'Salary'})
# Convert 'Salary' column from object/string to integer
df['Salary'] = df['Salary'].astype(int)
print(df.dtypes, "\n")
```

```python
# 6. Display random employee records for verification
print("--- 6. Random Employee Records ---")
# Display a sample of 3 random records
random_records = df.sample(n=3)
print(random_records, "\n")
```

```python
# 7. Find employees with the highest and lowest salaries
print("--- 7. Highest and Lowest Salaries ---")
highest_salary_emp = df[df['Salary'] == df['Salary'].max()]
lowest_salary_emp = df[df['Salary'] == df['Salary'].min()]
print("Highest Salary Employee(s):")
print(highest_salary_emp)
print("Lowest Salary Employee(s):")
print(lowest_salary_emp, "\n")
```

```python
# 8. Calculate employee bonuses using a custom function
print("--- 8. Calculating Bonuses ---")
def calculate_bonus(row):
    # Bonus is 10% for IT, 5% for others
    if row['Dept'] == 'IT':
        return row['Salary'] * 0.10
    else:
        return row['Salary'] * 0.05

df['Bonus'] = df.apply(calculate_bonus, axis=1)
print(df[['Emp_Name', 'Dept', 'Salary', 'Bonus']], "\n")
```

```python
# 9. Convert performance grades into meaningful descriptions
print("--- 9. Converting Performance Grades ---")
grade_mapping = {
    'A': 'Excellent',
    'B': 'Good',
    'C': 'Satisfactory',
    'D': 'Needs Improvement'
}
df['Performance_Description'] = df['Grade'].map(grade_mapping)
print(df[['Emp_Name', 'Grade', 'Performance_Description']], "\n")

print("--- Final Cleaned & Analyzed Dataset ---")
print(df)
```
